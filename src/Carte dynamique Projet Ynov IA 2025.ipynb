# Projet 2 Données cartographiques concernant l'épidémie de COVID-19

Ce projet consiste à créer des cartes à partir de Données cartographiques de l'épidémie de la COVID-19

L'information officielle sur la progression de l'épidémie en France a été au départ assez fragmentée. Différentes initiatives ont tenté de structurer celle-ci sous forme de données libres. Malgré ce travail, les données ont été néanmoins souvent difficilement exploitables à l'état brut au sein d'outils cartographiques. 

L'objectif de ce projet est de consolider l'information et de la rendre disponible dans des formats ouverts et aisément réutilisables pour produire des cartes. Le format pivot privilégié est le GeoJson.

Les données sont proposées selon plusieurs granularités : régions et départements. Les données à la maille du département ont été initialement fragmentaires, néanmoins sous l'impulsion des initiatives libres telles que OpenCovid19, des données plus précises relatives à l’épidémie ont été mises à disposition à disposition par Santé publique France.
 
 vous trouvez les donées dans le lien suivant:
https://www.data.gouv.fr/fr/datasets/donnees-cartographiques-concernant-lepidemie-de-covid-19/


Pour plus de détails ou la dernière version des données consulter 
https://github.com/kalisio/covid-19

Dans ce projet, Nous souhaions utiliser ces données (fournis sous format GeoJson) pour entrainer nos modèle d'apprentissage. 

1-	Classification non supervisé sur les données afin de créer des cluster homogènes.
2-	Visualisation des données sur des cartes
    --> une carte dynamique  et synchronisée est un poit fort appréciable.
    
3- Nous pouvons, pour rendre le travail plus pertinent, intégrer de nouvelles données relatives au vaccin. Le but ici est d'extraire des enseignements sur la correlation de la vaccination et la progression/l'évolution de la COVID19.
